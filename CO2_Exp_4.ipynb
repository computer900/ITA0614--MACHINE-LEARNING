```python
# Artificial Neural Network using Backpropagation

from sklearn.neural_network import MLPClassifier

# Training data (XOR)
X = [[0,0],[0,1],[1,0],[1,1]]
y = [0,1,1,0]

# Create and train ANN
model = MLPClassifier(hidden_layer_sizes=(2,),
                      activation='logistic',
                      solver='lbfgs',
                      max_iter=1000,
                      random_state=1)

model.fit(X, y)

# Test the model
print("Predictions:")
for i in X:
    print(i, "->", model.predict([i])[0])

# Test with a new sample
new_sample = [[1,0]]
print("\nNew Sample:", new_sample[0])
print("Prediction:", model.predict(new_sample)[0])
```
